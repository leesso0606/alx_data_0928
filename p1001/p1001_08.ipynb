```python
# 1. item_name 당 주문 개수를 출력하시오. 10개만 출력
#  몇번 주문 기록에 등장했나 (행의 개수)
# df.groupby('item_name').size().head(10)
df['item_name'].value_counts().head(10)
```

```python
# 2. item_name 당 주문 총량(quantity)을 출력하시오.
#  실제로 몇개를 샀는가
df.groupby('item_name')['quantity'].sum().sort_values(ascending=False)
```

```python
# 3. 가장 많이 주문한 item_name  10개를 출력하시오.
# df.groupby('item_name').size().sort_values(ascending=False).head(10)

# item_name 컬럼에서 각 값의 등장수를 세고, 많이 등장한 순서대로 정렬한 다음 위에서 10개를 가져온다.
# value.counts()가 자동으로 개수가 많은 순서로 정렬해준다
df['item_name'].value_counts().head(10)
```

```python
# # 4. 주문 당 평균 금액을 출력하시오
# df['item_price']=df['item_price'].str.strip().str.replace("$","").astype(float)
df.groupby('order_id')['item_price'].sum().mean()
```

```python

# 5. 한 주문에 50달러 이상 사용한 주문 번호(id) 출력하시오.
df['item_price']=df['item_price'].str.strip().replace("$","").astype(float)
filt=df.groupby('order_id')['item_price']>=10
df[filt]['order_id']

```

```python
# 6.Chicken Bowl을 2개 이상 구매한 주문서 출력하시오
filt=(df['item_name']=='Chicken Bowl') &(df['quantity']>=2)
df[filt]['order_id']
```
