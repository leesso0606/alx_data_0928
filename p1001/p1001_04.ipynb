```python
import pandas as pd
df=pd.read_excel('file/score.xlsx',index_col='지원번호')
df
```

```python
# nan이 있을때
# null이 없으면 소문자로 변경, 있으면 없음으로 변경
def cal_sw(lang):
    if pd.notnull(lang):
        return lang.lower()
    else: return '없음'

# def cal_sw(lang):
#     if pd.notnull(lang):
#         return lang.lower()
#     else: return '없음'

df['SW특기']=df['SW특기'].apply(cal_sw)
df
```
