```python
import pandas as pd
df=pd.read_excel('file/score.xlsx',index_col='지원번호')
```

```python
# 그룹핑 grouby.size,mean,max,min,count ...

# 그룹화 - 통계, 평균
# 학교가 -구로고, 디지털고 인 행만 출력가능
# df.groupby('학교').get_group('구로고')
# df.groupby('학교').get_group('디지털고')

# 중복제외 학교의 값을 알수있음
df['학교'].unique()
# 학교기준으로 그룹핑을 해서 크기 확인 가능
df.groupby('학교').size()
# 학교 기준으로 구로고, 디저털고의 개수를 알 수 있음
df.groupby('학교').size()['구로고'] 
df.groupby('학교').size()['디지털고'] 

# 학교 기준으로 키의 평균을 알 수 있음
df.groupby('학교')['키'].mean()
# 학교 기준으로 국어의 평균을 알 수 있음
df.groupby('학교')['국어'].mean()
# 학교 기준으로 국어,수학의 평균을 알 수 있음
df.groupby('학교')[['국어','수학']].mean()
# 학교 기준으로 국어,수학의 합계을 알 수 있음
df.groupby('학교')[['국어','수학']].sum()
```

```python
# 그룹핑을 한 후 개수 확인 / 같은 말
# df.groupby('학교').size()
# value_counts()가 자동으로 개수가 많은 순서로 정렬
df['학교'].value_counts()
```

```python
# 학교로 그룹핑 한 다음, 학년을 기준으로 나눌 수 있다
# df.groupby('학교')['학년'].value_counts()

# 학교로 그룹핑 한 다음, 학년을 기준으로 나눔->구로고 것만 알 수 있음
# df.groupby('학교')['학년'].value_counts().loc['구로고']
# 학교로 그룹핑 한 다음, 학년을 기준으로 나눔->디지털고 것만 알 수 있음
# df.groupby('학교')['학년'].value_counts().loc['디지털고']

# normalize 백분율로 표시 :원형 그래프
df.groupby('학교')['학년'].value_counts(normalize=True).loc['구로고']
```

```python
# 학년 별 키의 평균->정렬
df.groupby('학년')['키'].mean().sort_values()

# 학년 별 국어, 영어의 합->국어를 기준으로 정렬하라
df.groupby('학년')[['국어','영어']].sum().sort_values('국어')
# 학교 별 국어, 영어의 평균->영어 기준으로 정렬하라
df.groupby('학교')[['국어','영어']].mean().sort_values('영어')
```
