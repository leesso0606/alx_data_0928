```python
# 함수적용 :NaN을 없음으로 변경, 결측값은 소문자로 변경
def capChange(lang):
    if pd.notnull(lang):
        return lang.lower()
    else:
        return '없음'

df['SW특기']=df['SW특기'].apply(capChange)
df['SW특기']

# NaN은 소문자로 안 바뀌고 그대로, 있는 값만 소문자로 변경됨
# df['SW특기'].str.lower()
```

```python
import pandas as pd
df=pd.read_excel('file/score.xlsx',index_col='지원번호')
```

```python
# 1.합계 (과목 총점)컬럼을 추가하시오
df['합계']=df['국어']+df['영어']+df['수학']+df['사회']+df['과학']
# 2.평균 컬럼을 추가하시오
df['평균']=df['합계']/5
# 3.등급 컬럼을 추가해서 90점 이상 A, 80점 이상 B, 70점 이상 C, 60점 이상 D, 그 이하는 F
# 함수를 사용할것
# r 값을 받아 각 결과 값을 return한다.
def rank(r):
    if r>=90:
        return'A'
    elif r>=80:
        return'B'
    elif r>=70:
        return'C'
    elif r>=60:
        return'D'
    else: return'F'

# 평균 컬럼의 값을 하나씩 rank 함수의 r에 전달하고,
# 함수의 반환값을 등급 컬럼에 저장한다.
df['등급']=df['평균'].apply(rank)
df
```

```python
# 합계의 값을 비교하여 등수
# ascending=False:역순정렬, methon='min'->1.2.2.4
df['등수']=df['합계'].rank(ascending=False).astype(int)
```
