```python
import pandas as pd
df=pd.read_excel('file/score.xlsx',index_col='지원번호')
```

```python
import random
df['요일']=0
for i in range(8):
    df.iloc[i,9]=random.randint(0,6)
df
```

```python
# 0-일1-월2-화3-수4-목5-금6-토
# 함수를 이용하여 출력하시오

def day(d):
    if d==0: return '일요일'
    elif d==1: return '월요일'
    elif d==2: return '화요일'
    elif d==3: return '수요일'
    elif d==4: return '목요일'
    elif d==5: return '금요일'
    else: return '토요일'

df['요일']=df['요일'].apply(day)
# df
```

```python
data = {
    '번호':['1번','2번','3번','4번','5번','6번','7번','8번'],
    '이름':['홍길동','유관순','이순신','강감찬','김구','김유신','홍길자','홍길순'],
    '달러':['$19,700','$18,400','$16,800','$18,700','$18,800','$20,200','$18,800','$19,000'],
    '우대':[90,40,80,40,15,80,55,100]
}
df=pd.DataFrame(data)
# index를 번호로 하고 저장한다
df.set_index('번호',inplace=True)
df
```

```python
# 금액컬럼을 추가하시오(금액:달러+달러*우대*0.01)
df['금액']=0
# 우대 50 이상인 사람에게 달러+달러*우대*0.01
# 우대가 50미만은 달러만 입력
```

```python
# 달러가 str타입->int타입으로 변경($ , ,표시 제거)
df['달러']=df['달러'].str.replace('$','')
df['달러']=df['달러'].str.replace(',','').str.strip().astype(int)
```

```python
# 2개가 넘어와야 함. df['달러'], df['우대']
def cal_change(df2):
    if df2['우대']>=50:
        return df2['달러']+(df2['달러']*df2['우대']*0.01)
    else: return df2['달러']
df['금액']=df.apply(cal_change,axis=1)
df
```
