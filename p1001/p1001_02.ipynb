# index

```python
import pandas as pd
df=pd.read_excel('file/score.xlsx', index_col='지원번호')
```

```python
# row:index수정, 추가, 삭제
# index 삭제-조건문을 사용하여 삭제

# index값으로 삭제
# df.drop(index=['1번','2번'],inplace=True)


# 조건을 이용한 index삭제
# 수학점수가 80점 미만학생 삭제
filt=df['수학']<80
# df2=df[~filt]
# df2 #수학점수 80점 이상 학생만 출력
# df[filt].count() 몇명인지 알 수 있음

df.drop(index=df[filt].index,inplace=True)
df
```

```python
df.columns
# 국어,영어,수학,과학,사회,합계 컬럼을 추가하고 합계 점수가 350점 이상만 출력하시오

# 컬럼 추가
df['합계']=df['국어']+df['영어']+df['수학']+df['사회']+df['과학']
# 350점 이상만 출력:원본은 그대로 조회
filt=df['합계']>=350
df[filt]
# 350점 미만 삭제 후 출력:원본에서도 삭제
df.drop(index=df[~filt].index,inplace=True)
```

```python
# index추가:loc[]이용, 컬럼 개수만큼 입력
df.loc['9번']=['홍길동','구로고',190,90,80,70,60,50,'Python']

# 특정 위치 부분값을 추가 또는 수정:index, columns 지정하여 값 입력
df.loc['4번','SW특기']='C'

# 학교, SW특기 컬럼에 값 부분 수정
df.loc['4번',['학교','SW특기']]=['디지털고','Python']
# 
df.loc['5번',['학교','사회','SW특기']]=['디지털고',50,'Java']
```

```python
# 컬럼 이름 변경:전체수정
# df.columns=[변경할 이름]
# df2=df[['이름','국어','영어']]
# df2.columns=['name','kor','eng']

# 부분 컬럼 이름 변경
df.rename(columns={'이름':'name','국어':'kor','영어':'eng'},inplace=True)
df.rename(columns={'학교':'school'},inplace=True)
df
```
