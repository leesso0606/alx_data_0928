```python
import pandas as pd
df=pd.read_excel('file/score.xlsx', index_col='지원번호')
```

```python
# 문자열 함수 .str함수 - .str.startswith():('강')으로 시작하는 문자열
# startswith()-시작,endswith()-끝,contains()-포함,lower()-소문자,upper()-대문자,capitalize()-첫글자만 대문자로
# strip()-공백제거, split()-분리, replace()-대체, len()-글자 수, find()-찾기

# startswith():('강')으로 시작하는 문자열
# df[df['이름'].str.startswith('강')]
df.loc[df['이름'].str.startswith('강')]

# contains():('원')이 포함되어 있는 글자
# 문자열이 포함되어 있는지
df[df['이름'].str.contains('원')]
df.loc[df['이름'].str.contains('원')]

# ~는 not :원이 포함 안되어 있는 글자
df[~df['이름'].str.contains('원')]
df.loc[~df['이름'].str.contains('원')]

# lower,upper,capitalize
df['SW특기'].str.lower() #소문자로 변경
df['SW특기'].str.upper() #대문자로 변경
df['SW특기'].str.capitalize() #첫 글자만 대문자로 변경

# isin :여러개 검색['python','java'](.str없이 )
# 값이 목록과 같은지
# SW특기를 소문자로 바꾼 다음, python 또는 java와 정확히 일치하는 값을 찾는다.
langs=['python','java']
df[df['SW특기'].str.lower().isin(langs)] # SW특기 값 소문자로 변경 후 리스트 검색


# df['SW특기'].str.lower().contains()
df['SW특기']
df['SW특기'].str.contains('java',na=False) # na=False :NaN 데이터 출력 안됨 /NaN을 False로 처리
df[df['SW특기'].str.contains('java',na=True)]# na=True :NaN 데이터 출력 됨 /NaN을 True로 처리
df[df['SW특기'].str.lower().str.contains('java',na=False)] #소문자로 변경 후 검색

# unique :중복제거
df['학교'].unique()
```
