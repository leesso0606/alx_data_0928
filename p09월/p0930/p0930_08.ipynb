```python
df=pd.read_csv('file/jumin08_2.csv',index_col='행정구역')

# 두 개의 컬럼만 가져와서 df2를 만듬
df2=df[['2026년08월_총인구수','2026년08월_세대수']]

# ,를 제거하고 int로 변환
df2['2026년08월_총인구수']=df2['2026년08월_총인구수'].str.replace(',','').astype(int)
df2['2026년08월_세대수']=df2['2026년08월_세대수'].str.replace(',','').astype(int)

# 두 컬럼의 데이터 타입 등을 확인
df2.info()

# 총인구수를 기준으로 오름차순 정렬 (작은 값 → 큰 값)
df2.sort_values('2026년08월_총인구수')
```
