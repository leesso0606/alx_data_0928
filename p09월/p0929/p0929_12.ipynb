```python
import pandas as pd
df=pd.read_excel('file/합계출산율_2026.xlsx',skiprows=28,nrows=2)
df.set_index('Unnamed: 0',inplace=True)
df.index.name='제목'
df
```

```python
df.index.values[0]
```

```python
# index의 이름 변경
df.rename(index={'출생아\xa0수':'출생아 수'},inplace=True)
df.rename(index={'합계\xa0출산율':'합계 출산율'},inplace=True)
```

```python
# 출생아 수가 250이상 300이하인 년도를 구하시오
df.loc['출생아 수'][(df.loc['출생아 수']>250)&(df.loc['출생아 수']<300)].astype(float)
# 최소, 평균을 구하시오
df.loc['출생아 수'].min().astype(float)
# # 평균
df.loc['출생아 수'].mean().astype(float)
```
