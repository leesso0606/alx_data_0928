```python
import pandas as pd
import numpy as np

df=pd.read_excel('file/score.xlsx',index_col='지원번호')
```

```python
# Nan데이터 처리
# fillna:nan 자리에 ()안에 있는 글자로 데이터를 채워줌

# df 전체의 모든 NaN을 '없음'으로 바꿈
# df.fillna('없음',inplace=True)
# df

# np.nan :numpy명령어 ,nan 데이터를 모두 입력
# numpy는 숫자만 가능, 문자가 안됨.
# np.nan을 컬럼에 넣으면 해당 컬럼의 자료형이 float64가 될 수 있음

df['학교']=np.nan #pandas최신버전 : float64타입변경됨.
# df['학교']

# # 문자열을 저장할 수 있도록 object 타입으로 변경 :object  → 범용적인 객체 자료형(문자열, 정수, 실수, nan(결측값)을 다 넣을 수 있는 상자)
df['학교']=df['학교'].astype('object')

# # NaN을 '없음'으로 채우고 원본 df에 반영
df.fillna('없음',inplace=True)
# # df
```
