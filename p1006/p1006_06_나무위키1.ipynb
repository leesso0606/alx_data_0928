```python
# 3. 파일 BeautifulSoup변환
with open('file/namu1.html','r',encoding='utf-8') as f:
    soup = BeautifulSoup(f,'lxml')
```

```python
# title:'조원우'
# url:https://namu.wiki/w/%ED%8D%BD(%EA%B2%8C%EC%9E%84%20%EC%9A%A9%EC%96%B4) 식으로 저장 예정
```

```python
datas=soup.find('div',{'class':'vR3FnPgQ'})
divs=datas.find_all('div',{'class':'M5XHy6jV BqyggfHC'})
namu_data=[]
for i in range(len(divs)):
    a_tag=divs[i].find('a')
    title=a_tag.get_text(strip=True) if a_tag else None
    # print(title)
    url=a_tag['href']
    urls=f'https://namu.wiki{url}'
    # print(urls)

    namu_data.append({
        'title':title,
        'url':urls
    })
```

```python
# 선생님이 하신것
datas = soup.find('div',{'class':'vR3FnPgQ'})
divs = datas.find_all('div',{'class':'M5XHy6jV BqyggfHC'})
len(divs)

namu_data = []
for div in divs:
    # 문서명
    title_tag = div.find('a')
    title = title_tag.get_text(strip=True) if title_tag else None
    # print(title)
    first_url = 'https://namu.wiki'
    # url = title_tag['href']
    url = first_url+title_tag['href']
    # print(url)
    namu_data.append({
        'title':title,
        'url':url
    })
```

```python
# dataFrame으로 변환해서 저장하시오. #namu_link.csv
df=pd.DataFrame(namu_data)
df.to_csv('file/namu_link.csv')
df
```
