```python
import matplotlib.pyplot as plt
import matplotlib
matplotlib.rcParams['axes.unicode_minus']=False    # -가 안깨지도록(- 부호 처리)
matplotlib.rcParams['font.family']='Malgun Gothic'  #한글처리: 윈도우 맑은 고딕 저장
matplotlib.rcParams['font.size']=15                 #전체 글자폰트크기 조정
# matplotlib.rcParams['font.family']='AppleGithic'    #맥컴 애플고딕 저장

import pandas as pd
import numpy as np #NumPy는 숫자 계산용
```

```python
df=pd.read_excel('file/students.xlsx')
# 학년별로 그룹핑->백분률
school=df['학교'].value_counts(normalize=True)
explode=[0.1,0,0] #원파트 띄우기
school.index
school.values
plt.pie(school.values,labels=school.index,startangle=90,counterclock=False,explode=explode,\
        autopct='%.1f%%') #퍼센트를 소수점 첫째 자리까지 표시해라
plt.show()
```

```python
# values=[1,1,1,1]
values=[30,25,20,13,10,2] #자동으로 백분율로 계산
labels=['파이썬','html','자바스크립트','판다스','그래프','디장고']
explode=[0.2,0.1,0,0,0,0] #원파트 띄우기
colors = ['#ffadad','#ffd6a5','#fdffb6','#caffbf','#9bf6ff','#a0c4ff']

# startangle 기준각도, counterclock:시계방향
# autopct='%.1f%%':퍼센트를 소수점 첫째 자리까지 표시해라
plt.pie(values,labels=labels,startangle=90,counterclock=False,explode=explode,\
        autopct='%.1f%%',colors=colors) 
plt.title('언어별 선호도')
plt.legend(loc=(1.2,0.3),title='언어별 선호도') #범례
plt.show()
```

```python
# 한국 막대 그래프를 red추가해서 출력하시오
df=pd.read_csv('file/drinks.csv')
df['total_servings']=df['beer_servings']+df['spirit_servings']+df['wine_servings']
#  순위 높은 순으로 정렬
sort_df=df.sort_values('total_servings',ascending=False)
# 순위 높은 순부터 정렬해서, 합계에 대한 막대 그래프를 출력하시오
x=sort_df['country']
y=sort_df['total_servings']

k_sort_df=sort_df[sort_df['country']=='South Korea'] #South Korea인 행만 골라냄
x2=k_sort_df['country']
y2=k_sort_df['total_servings']


plt.bar(x,y,label='국가별')
plt.bar(x2,y2,color='red',label='한국')
plt.legend()
plt.show()
```
