```python
df=pd.read_csv('file/namu2.csv')
df=df.head(5)
```

```python
# 글자에 대한 전처리 작업-한글이외 글자 삭제, 필요없는 글자 삭제
def text_cleaning(text):
    if pd.isna(text):
        return''
    hangul = re.compile('[^ ㄱ-ㅣ가-힣]+')  #정규표현식에 해당이 되는지 확인
    return hangul.sub('',text)
```

```python
df['content']=df['content'].apply(lambda x: text_cleaning(x))
df['content']
# df_content_all=
```

```python
# python -m pip install konlpy 설치

from konlpy.tag import Okt #한글 형태소
from collections import Counter
import os
os.environ["JAVA_HOME"] = r"C:\jdk21"
```

```python
n_tag=Okt()
nouns=n_tag.nouns(df['content'][0]) #명사로 추출
count=Counter(nouns) # 명사로 추출한 단어 개수 확인
count
```

```python
# 1개 단어 키워드 제외-딕셔너리 타입
# count #딕셔너리 타입
remove_counter=Counter({x:count[x] for x in count if len(x)>1})
# 위랑 같은말
# for x in count:
#     if len(x)>1:
#         {x:count[x]}
remove_counter
```
