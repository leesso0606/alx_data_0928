```python
import matplotlib.pyplot as plt
import matplotlib
matplotlib.rcParams['axes.unicode_minus']=False    # -가 안깨지도록(- 부호 처리)
matplotlib.rcParams['font.family']='Malgun Gothic'  #윈도우 맑은 고딕 저장
matplotlib.rcParams['font.size']=15                 #전체 글자폰트크기 조정
# matplotlib.rcParams['font.family']='AppleGithic'    #맥컴 애플고딕 저장

import pandas as pd
import numpy as np
```

```python
df=pd.read_excel('file/score.xlsx',index_col='지원번호')
```

```python
df.loc['1번','국어']
df.loc['1번','영어']
df.loc['1번','수학']
```

```python
# 선 스타일
x = df['이름']
y = df['국어']
y2 = df['영어']
y3= df['수학']
plt.figure(figsize=(10,5))
colors=['red','green','blue']

plt.title('학생성적 프로그램')
plt.bar(x,y,label='국어',width=0.5)
plt.bar(x,y2,bottom=y,label='영어',width=0.5)
plt.bar(x,y3,bottom=y+y2,label='수학',width=0.5)
# ncol=3 범례가 1줄로 나타남
plt.legend(ncol=3)
plt.grid(axis='y',linestyle=':',alpha=0.7)
for i,(v,v2,v3) in enumerate(zip(y,y2,y3)):
    plt.text(i,v/2,v,ha='center',va='center')
    plt.text(i,v+v2/2,v2,ha='center',va='center')
    plt.text(i,v+v2+v3/2,v3,ha='center',va='center')



plt.show()
```
