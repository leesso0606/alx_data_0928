```python
import matplotlib.pyplot as plt
import matplotlib
matplotlib.rcParams['axes.unicode_minus']=False    # -가 안깨지도록(- 부호 처리)
matplotlib.rcParams['font.family']='Malgun Gothic'  #윈도우 맑은 고딕 저장
matplotlib.rcParams['font.size']=15                 #전체 글자폰트크기 조정
# matplotlib.rcParams['font.family']='AppleGithic'    #맥컴 애플고딕 저장

import pandas as pd
import numpy as np #NumPy는 숫자 계산용
```

```python
# student.xlsx 파일을 읽어와서
# 학교별 국어, 영어, 수학 평균을 막대 그래프로 출력하시오

# 학교별, 학년별 국어,영어,수학 평균을 출력하시오
# 합계,평균,등수 컬럼을 추가하고 값을 입력하시오
```

```python
df=pd.read_excel('file/students.xlsx',index_col='번호')
```

```python
df.info()
df.describe()
```

```python
# 학교 별 과목 평균
kor=df.groupby('학교')['국어'].mean().round(1)
eng=df.groupby('학교')['영어'].mean().round(1)
math=df.groupby('학교')['수학'].mean().round(1)

# 학교 별 과목 평균-막대 그래프 출력
x=np.arange(3) #arange는 연속된 숫자를 만들어주는 함수
y=kor
y2=eng
y3=math

plt.title('학교별 성적 평균')

plt.bar(x-0.2,y,label='국어',width=0.2)
plt.bar(x,y2,label='영어',width=0.2)
plt.bar(x+0.2,y3,label='수학',width=0.2)

for i, (v,v2,v3)in enumerate(zip(y,y2,y3)):
    plt.text(i-0.2,v+1,v,ha='center')
    plt.text(i,v2+1,v2,ha='center')
    plt.text(i+0.2,v3+1,v3,ha='center')

plt.legend(ncol=3)
plt.xticks(x,['구로고', '디지털고','신림고'])
plt.show()


```

```python
# 선생님이 하신것
df = pd.read_excel('file/students.xlsx',index_col='번호')
# 학교별 국어,영어,수학평균을 막대그래프로 출력하시오.
school = df.groupby('학교')[['국어', '영어', '수학']].mean()
school
```

```python

x = school.index
index = np.arange(school.shape[0])
y = school['국어']
y2 = school['영어']
y3 = school['수학']

plt.title('학교별 성적그래프')
plt.bar(index-0.2,y,label='국어',width=0.2)
plt.bar(index,y2,label='영어',width=0.2)
plt.bar(index+0.2,y3,label='수학',width=0.2)
plt.xticks(index,school.index)
plt.yticks([0,50,80,100])
for i,(v,v2,v3) in enumerate(zip(y,y2,y3)):
    plt.text(i-0.2,v+2,f'{v:.1f}',ha='center')
    plt.text(i,v2+2,f'{v2:.1f}',ha='center')
    plt.text(i+0.2,v3+2,f'{v3:.1f}',ha='center')

plt.show()
```

```python
# 학교별, 학년별 국어,영어,수학 평균을 출력하시오.
```

```python
df=pd.read_excel('file/students.xlsx',index_col='번호')

kor2=df.groupby(['학교','학년'])['국어'].mean()
eng2=df.groupby(['학교','학년'])['영어'].mean()
math2=df.groupby(['학교','학년'])['수학'].mean()
kor2
```

```python
# 학교별, 학년별 국어,영어,수학 평균을 출력하시오.
result = df.groupby(['학교', '학년'])[['국어', '영어', '수학']].mean()
result
```

```python
# 각 학교별->학년별 국,영,수 평균을 구하시오
kor2=df.groupby(['학교','학년'])['국어'].mean().round(1)
eng2=df.groupby(['학교','학년'])['영어'].mean().round(1)
math2=df.groupby(['학교','학년'])['수학'].mean().round(1)

# 학교 별 과목 평균-막대 그래프 출력
x=np.arange(len(kor2)) #arange는 연속된 숫자를 만들어주는 함수
y=kor2
y2=eng2
y3=math2

plt.figure(figsize=(15,5))

plt.bar(x-0.2,y,width=0.2,label='국어')
plt.bar(x,y2,width=0.2,label='영어')
plt.bar(x+0.2,y3,width=0.2,label='수학')

for i ,(v,v2,v3) in enumerate (zip(y,y2,y3)):
    plt.text(i-0.2,v+1,v,ha='center')
    plt.text(i,v2+1,v,ha='center')
    plt.text(i+0.2,v3+1,v,ha='center')

plt.legend(ncol=3)
plt.xticks(x,['구로고 1','구로고 2','구로고 3','디지털고 1','디지털고 2','디지털고 3','신림고 1','신림고 2','신림고 3'])



plt.show()
```

```python
# 합계,평균,등수 컬럼을 추가하고 값을 입력하시오
df['합계']=df['국어']+df['영어']+df['수학']
df['평균']=df['합계']/3
df['등수']=df['합계'].rank(ascending=False).astype(int)
```
