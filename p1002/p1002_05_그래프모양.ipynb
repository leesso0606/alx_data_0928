```python
import matplotlib.pyplot as plt
import matplotlib
matplotlib.rcParams['axes.unicode_minus']=False    # -가 안깨지도록(- 부호 처리)
matplotlib.rcParams['font.family']='Malgun Gothic'  #윈도우 맑은 고딕 저장
matplotlib.rcParams['font.size']=15                 #전체 글자폰트크기 조정
# matplotlib.rcParams['font.family']='AppleGithic'    #맥컴 애플고딕 저장

import pandas as pd
import numpy as np
```

```python
# 선 스타일
x = [1,2,3]
y = [2,4,8]
plt.title('꺾은선 graph')  # 상단타이틀

# figure:그래프 사이즈 qusrud, figsize:크기, dip:해상도
# facecolor:배경색 변경
# plt.figure(figsize=(20,5))
# plt.figure(figsize=(15,2), dpi=200,facecolor='yellow')
plt.plot(x,y,linewidth=2,marker='o',ms=10, mfc='yellow',linestyle='-.',\
        color='blue')

# 그래프 안에 글자를 입력:plt.text()
# plt.text(x[0],y[0]+0.3,2,ha='center',color='red')
for idx,value in enumerate(y):
        plt.text(x[idx]-0.05,y[idx]+0.2,value,ha='center',color='red')

# 그래프 간격조정 ( 입력값 형태로 구성 )
# plt.xticks([1,1.2,1.4,1.6,2,3])
# plt.yticks([-15,10,20,30]) #간격부분 조정을 하면 꺾이는 상태를 조정
# plt.yticks([-15,10,20,30,100]) #간격부분 조정을 하면 꺾이는 상태를 조정
# plt.savefig('file/aaa2.png',dpi=200) #스크린 샷 저장
plt.show()
```

```python
# 선 스타일
x = [1,2,3]
y = [20,40,80]
y2=[50,10,30]
y3=[10,20,50]


plt.title('꺾은선 graph')  # 상단타이틀

# figure:그래프 사이즈 qusrud, figsize:크기, dip:해상도
# plt.figure(figsize=(15,2), dpi=200,facecolor='yellow')
plt.plot(x,y,label='국어',linewidth=2,marker='o',ms=10, mfc='yellow',linestyle='-',\
        color='pink')
plt.plot(x,y2,label='영어',linewidth=2,marker='x',ms=10, mfc='yellow',linestyle=':',\
        color='blue')
plt.plot(x,y3,label='수학',linewidth=2,marker='^',ms=10, mfc='yellow',linestyle='-.',\
        color='green')
plt.legend()

# 그래프 안에 글자를 입력:plt.text()
for index,value in enumerate(y):
        plt.text(x[index],y[index]+2,value,ha='center',color='red')
for index,value in enumerate(y2):
        plt.text(x[index],y2[index]+2,value,ha='center',color='red')
for index,value in enumerate(y3):
        plt.text(x[index],y3[index]+2,value,ha='center',color='red')



# 그래프 간격조정 ( 입력값 형태로 구성 )
# plt.xticks([1,1.2,1.4,1.6,2,3])
# plt.yticks([-15,10,20,30,100]) #간격부분 조정을 하면 꺾이는 상태를 조정

# 그래프 저장/스크린 샷 저장
plt.savefig('file/aaa.png')
plt.savefig('file/aaa2.png',dpi=200)
plt.show()
```

```python
# 선 스타일
x = [1,2,3]
# x = ['홍길동','유관순','강감찬']
y = [20,40,80]
y2=[50,10,30]
y3=[10,20,50]
colors=['red','green','blue']

plt.title('학생성적 프로그램')

# color=그래프 색변경.  1개('red) 2개 ['red','green]
# width:막대그래프 넓이 조절. alpth:투명도, 
bar=plt.bar(x,y,color=colors,width=0.8,alpha=0.5)
bar[0].set_hatch('/')
bar[1].set_hatch('x')
bar[2].set_hatch('..')

# grid-격자선 , 안내선
plt.grid(axis='y',linewidth=2,color='purple',linestyle='-.',alpha=0.5)
plt.grid(axis='x')
# 가로 막대 그래프
# plt.barh(y,x)
# rotation :각도조절
# plt.xticks(x,rotation=90)
for i,v in enumerate(y):
    plt.text(x[i],y[i],v,ha='center')

plt.show()
```
