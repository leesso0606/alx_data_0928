```python
# 선 스타일
x = [1,2,3]
y = [20,40,80]
y2=[50,10,30]
y3=[10,20,50]
colors=['red','green','blue']

plt.title('학생성적 프로그램')

# color=그래프 색변경.  1개('red) 2개 ['red','green]
# width:막대그래프 넓이 조절. alpth:투명도, 
bar=plt.bar(x,y,color=colors,width=0.8,alpha=0.5)
bar[0].set_hatch('/')
bar[1].set_hatch('x')
bar[2].set_hatch('..')

# grid
plt.grid(axis='y',linewidth=2,color='purple',linestyle='-.',alpha=0.5)
plt.grid(axis='x')
# 가로 막대 그래프
# plt.barh(y,x)
plt.xticks(x,rotation=45)
for i,v in enumerate(y):
    plt.text(x[i],y[i],v,ha='center')

plt.show()
```
