```python
import matplotlib.pyplot as plt
import matplotlib
matplotlib.rcParams['axes.unicode_minus']=False    # -가 안깨지도록(- 부호 처리)
matplotlib.rcParams['font.family']='Malgun Gothic'  #윈도우 맑은 고딕 저장
matplotlib.rcParams['font.size']=15                 #전체 글자폰트크기 조정
# matplotlib.rcParams['font.family']='AppleGithic'    #맥컴 애플고딕 저장
```

```python
import pandas as pd
import numpy as np
df=pd.read_csv('file/chipotle.tsv',sep='\t')
```

```python
# 가장 주문이 많이 된 메뉴의 총 개수를 출력하시오
# 주문서 2개,3개 ->총 몇개가 주문이 들어 왔는지 출력하시오
# 막대그래프로 출력하시오
```

```python
item_graph=df.groupby('item_name')['quantity'].sum()
x=item_graph.index
y=item_graph.values

plt.bar(x,y)
plt.title('메뉴별 주문수량')
plt.show()
```
